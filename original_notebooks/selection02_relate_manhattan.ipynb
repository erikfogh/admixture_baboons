Augmenting with runstats

```python
from scipy.stats import chi2
def fisher_method(pvalues):
    df = 2* len(pvalues)
    return 1-chi2.cdf(-2*sum(map(np.log, pvalues)), df)
```

```python
relate_sele = pd.read_csv("../results/Tanzania_Olive_relate/chromhapX_selection.sele", sep=" ")
relate_sele["-log10 P-value"] = -relate_sele["when_mutation_has_freq2"]
relate_sele["p-relate"] = np.power(10, relate_sele["when_mutation_has_freq2"])
runstats_sele = pd.read_hdf('../results/runstats/Eastern_Yellow_relate_chromhapX_runstats.h5')
```

```python
runstats_sele #.pos.astype(int)
```

```python
nr_run_subset = runstats_sele.loc[runstats_sele.stat_name == "nr_runs"]
```

```python
nr_run_mapping = dict(zip(nr_run_subset.pos.astype(int),(nr_run_subset.p)))
```

```python
relate_sele["nr_run_p"] = relate_sele.pos.map(nr_run_mapping)
```

```python
relate_sele[["p-relate", "nr_run_p"]]
```

```python
import itertools
n = 100
x = np.linspace(1e-10, 1, n)
a = np.array(list(itertools.product(x, x)))
b = np.apply_along_axis(fisher_method, 1, relate_sele[["p-relate", "nr_run_p"]])
c = b.reshape(n, n)
data = pd.DataFrame(c)
```

```python
relate_sele["combined_p"] = np.apply_along_axis(fisher_method, 1, relate_sele[["p-relate", "nr_run_p"]])
```

```python
relate_sele["log_combined_p"] = -np.log10(relate_sele["combined_p"])
```
